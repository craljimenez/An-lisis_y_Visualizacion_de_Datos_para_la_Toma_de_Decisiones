## 3. Operadores

Los operadores permiten hacer cálculos y comparaciones. Son la base de cualquier cálculo financiero (interés, IVA, utilidad) y de cualquier regla de negocio del estilo "si las ventas superan la meta y el margen es positivo, entonces...". Por ejemplo: calcular el valor futuro de una inversión con interés compuesto, comparar el gasto real contra el presupuesto (`gasto_real > presupuesto`), o sumarle el IVA al total de una factura.

**Aritméticos:**

| Operador | Operación | Ejemplo |
|---|---|---|
| `+` | Suma | `5 + 3` → `8` |
| `-` | Resta | `5 - 3` → `2` |
| `*` | Multiplicación | `5 * 3` → `15` |
| `/` | División | `10 / 4` → `2.5` |
| `**` | Potencia | `2 ** 3` → `8` |
| `%` | Residuo de una división (módulo) | `10 % 3` → `1` |

**Relacionales** (siempre devuelven `True` o `False`):

| Operador | Operación | Ejemplo |
|---|---|---|
| `==` | Igual a | `5 == 5` → `True` |
| `!=` | Diferente de | `5 != 3` → `True` |
| `>` | Mayor que | `5 > 3` → `True` |
| `<` | Menor que | `5 < 3` → `False` |
| `>=` | Mayor o igual que | `5 >= 5` → `True` |
| `<=` | Menor o igual que | `5 <= 3` → `False` |

**Lógicos** (combinan varias condiciones a la vez):

| Operador | Operación | Ejemplo |
|---|---|---|
| `and` | Verdadero solo si ambas condiciones son verdaderas | `True and False` → `False` |
| `or` | Verdadero si al menos una condición es verdadera | `True or False` → `True` |
| `not` | Invierte el valor de la condición | `not True` → `False` |

Además existe una forma abreviada muy usada para actualizar una variable a partir de su propio valor: `variable += valor` significa exactamente lo mismo que `variable = variable + valor`. Funciona igual con `-=`, `*=` y `/=`, y se usa todo el tiempo para ir acumulando un total dentro de un ciclo, algo que veremos más adelante en la clase.

## 4. Indexación: cómo acceder a una posición dentro de una secuencia

Muchos tipos de datos en Python son secuencias ordenadas de elementos: un texto (`str`) es una secuencia de caracteres, y más adelante veremos que las listas y las tuplas también lo son. En cualquier secuencia, cada elemento tiene una posición, llamada índice, y se accede a ella con corchetes: `secuencia[indice]`.

La indexación en Python siempre empieza en 0, no en 1: el primer elemento está en la posición 0, el segundo en la posición 1, y así sucesivamente. También se puede indexar desde el final usando números negativos: `-1` es el último elemento, `-2` el penúltimo, `-3` el antepenúltimo, etc.

Además de pedir un solo elemento, se puede pedir un pedazo de la secuencia (una subcadena, si es texto) con la sintaxis `secuencia[inicio:fin]`. El elemento en la posición `inicio` sí se incluye, pero el que está en la posición `fin` no. Si omites `inicio`, Python empieza desde el 0; si omites `fin`, llega hasta el final de la secuencia.

````python
ciudad = "Medellín"

# Indexación positiva: contando desde el inicio, empezando en 0
print(ciudad[0])    # 'M', primer caracter
print(ciudad[3])    # 'e', cuarto caracter

# Indexación negativa: contando desde el final
print(ciudad[-1])   # 'n', último caracter
print(ciudad[-2])   # 'í', penúltimo caracter

# Subcadenas con slicing: secuencia[inicio:fin]
print(ciudad[0:4])  # 'Mede'
print(ciudad[:4])   # 'Mede', si omites el inicio, arranca en 0
print(ciudad[4:])   # 'llín', si omites el fin, llega hasta el final
print(ciudad[-4:])  # 'llín', el mismo resultado usando índices negativos
````

Este mismo mecanismo es muy usado para extraer partes de un código o de un identificador de negocio. Por ejemplo, si el código de un producto siempre trae el año incluido en las mismas posiciones, se puede extraer con slicing en lugar de tener que leerlo a mano.

````python
codigo_producto = "COL-2024-00875"

anio = codigo_producto[4:8]
consecutivo = codigo_producto[9:]

print("Año del código:", anio)
print("Consecutivo:", consecutivo)
````

## 5. Listas

Una lista es una colección ordenada de valores, escrita entre corchetes `[ ]`, por ejemplo `[820, 950, 700]`. Cada valor ocupa una posición dentro de la lista y se indexa con la misma lógica que vimos con los textos: `lista[0]` trae el primer elemento, `lista[1]` el segundo, y así sucesivamente; `lista[-1]` trae el último elemento y `lista[-2]` el penúltimo. También se puede tomar un pedazo de la lista con slicing, `lista[inicio:fin]`, que devuelve una sublista con los elementos desde la posición `inicio` hasta la posición `fin - 1`, exactamente igual que al sacar una subcadena de un texto.

A diferencia de un texto, una lista sí se puede modificar después de creada: se le puede cambiar un valor asignándolo directamente sobre su posición (`lista[0] = nuevo_valor`), o agregarle un elemento nuevo al final con el método `.append(valor)`.

Además, existen varias funciones que se aplican directamente sobre una lista completa: `len(lista)` cuenta cuántos elementos tiene, `sum(lista)` suma todos sus valores, `max(lista)` y `min(lista)` devuelven el mayor y el menor, y `sorted(lista)` devuelve una copia ordenada de menor a mayor.

Cualquier serie de valores repetidos, ventas diarias, precios de una canasta de productos, calificaciones de un curso, se puede representar como una lista antes de pasar a estructuras más avanzadas como los DataFrames de Pandas, que veremos en el Módulo 2. Piensa en la lista de precios del dólar en la última semana, la de salarios de un equipo, o la de montos vendidos por cada vendedor del mes.

````python
# Ventas diarias de una tienda (en miles de pesos), de lunes a domingo
ventas_semana = [820, 950, 700, 1100, 990, 1500, 1330]

# Indexación, igual que con un texto
print("Venta del primer día (índice 0):", ventas_semana[0])
print("Venta del tercer día (índice 2):", ventas_semana[2])
print("Venta del último día (índice -1):", ventas_semana[-1])
print("Ventas de fin de semana (sábado y domingo):", ventas_semana[5:7])

# Funciones útiles sobre listas
print("Número de días registrados:", len(ventas_semana))
print("Total de la semana:", sum(ventas_semana))
print("Venta más alta:", max(ventas_semana))
print("Venta más baja:", min(ventas_semana))
print("Ventas ordenadas de menor a mayor:", sorted(ventas_semana))

# A diferencia de un texto, una lista sí se puede modificar
ventas_semana[0] = 900          # corregimos el registro del primer día
ventas_semana.append(1250)      # agregamos un nuevo día a la lista
print("Lista modificada:", ventas_semana)
````

## 6. Tuplas y su diferencia con las listas

Una tupla es, igual que una lista, una colección ordenada de valores que se indexa de la misma forma (índices positivos, negativos y slicing). La diferencia clave es que una tupla es inmutable: una vez creada, no se puede modificar, agregar ni quitar elementos. Se escribe entre paréntesis `( )` en lugar de corchetes.

Se usa una tupla cuando los datos representan algo que no debería cambiar accidentalmente: las coordenadas de una sede, un color en formato RGB, o un registro que llega de otro sistema y que no queremos alterar por error mientras lo procesamos. Cuando sí se espera que la colección cambie con el tiempo (agregar ventas, quitar clientes), la lista sigue siendo la opción natural.

````python
# Coordenadas de una sede en Medellín (latitud, longitud): un dato que no debería cambiar
coordenadas_tienda = (6.2442, -75.5812)

print(coordenadas_tienda[0])   # latitud
print(coordenadas_tienda[1])   # longitud
print(coordenadas_tienda[-1])  # también admite índices negativos

# Un registro de venta como tupla: fecha, producto, monto
registro = ("2024-03-05", "Mochila", 320000)
print(registro[1:])  # slicing: producto y monto
````

````python
# Ejecuta esta celda para ver el error real de intentar modificar una tupla
coordenadas_tienda = (6.2442, -75.5812)
coordenadas_tienda[0] = 10.0
````

El error `TypeError: 'tuple' object does not support item assignment` confirma justamente eso: una tupla no admite que le cambien un elemento después de creada. Con una lista, en cambio, `coordenadas_tienda[0] = 10.0` habría funcionado sin problema, tal como vimos en la sección anterior.

## 7. Diccionarios

Un diccionario (`dict`) es otra colección más, pero a diferencia de una lista o una tupla, no se indexa por posición (0, 1, 2...) sino por **clave**: cada valor se guarda asociado a un nombre que tú eliges. Se escribe entre llaves `{ }`, como pares `clave: valor`, por ejemplo `{"Cuaderno": 8500, "Mochila": 320000}`. Para leer un valor se usa `diccionario["clave"]`, no un número de posición.

Un diccionario sí se puede modificar después de creado: `diccionario["clave"] = valor` agrega una clave nueva si no existía, o actualiza su valor si ya existía. También se puede partir de un diccionario vacío `{}` e irle agregando claves a medida que se necesiten.

Para revisar si una clave ya existe se usa `in` (o `not in`), igual que se usó antes para buscar un valor dentro de una lista.

Para recorrer un diccionario con `for` hay tres variantes, según qué necesites en cada vuelta: `.keys()` entrega solo las claves, `.values()` entrega solo los valores, y `.items()` entrega la clave y el valor juntos (`for clave, valor in diccionario.items():`).

Cualquier dato que se identifique por nombre en vez de por posición es un buen candidato para un diccionario: el precio de cada producto de un catálogo, el inventario de cada bodega, el salario de cada empleado, todos identificados por su nombre en lugar de por el orden en que aparecen.

````python
# Precio de cada producto, identificado por su nombre (no por una posición)
precios = {"Cuaderno": 8500, "Mochila": 320000, "Lapicero": 1500}

print(precios["Mochila"])            # acceder por clave
print("Calculadora" in precios)      # ¿ya existe esa clave? (todavía no)
````

````python
# Modificar el diccionario: agregar una clave nueva y actualizar una existente
precios["Calculadora"] = 210000      # agregar una clave nueva
precios["Cuaderno"] = 9000           # actualizar el valor de una clave existente

print(precios)
````

````python
# Recorrer solo las claves, con .keys()
for producto in precios.keys():
    print(producto)
````

````python
# Recorrer solo los valores, con .values()
for precio in precios.values():
    print(precio)
````

````python
# Recorrer clave y valor a la vez, con .items()
for producto, precio in precios.items():
    print(producto, "cuesta", precio)
````

````python
# También se puede partir de un diccionario vacío e irlo llenando clave por clave
comisiones = {}

comisiones["Ana"] = 120000
comisiones["Luis"] = 95000

print(comisiones)
````

## 8. Funciones

Una función agrupa código que se repite, le da un nombre y permite reutilizarlo con distintos valores de entrada (parámetros). Se define con `def nombre_funcion(parametros):` y opcionalmente devuelve un resultado con `return`.

Los parámetros de una función pueden ser obligatorios u opcionales. Un parámetro es obligatorio cuando en la definición de la función no tiene un valor por defecto: hay que dárselo siempre al llamar la función, y si falta, Python responde con un `TypeError` señalando cuál hizo falta. Un parámetro es opcional cuando en su definición ya trae un valor por defecto, escrito como `parametro=valor`: si no lo mencionas al llamar la función, Python usa ese valor por defecto, y si lo mencionas, tu valor lo reemplaza.

Al llamar una función, los parámetros se pueden indicar por posición (en el mismo orden en que se definieron) o por nombre, escribiendo `nombre_parametro=valor`; llamándolos por nombre ya no importa el orden en que los escribas, y el código además queda más claro sobre qué representa cada valor.

Tiene sentido escribir una función cuando una empresa aplica la misma fórmula a muchos casos: calcular el IVA de cualquier venta, clasificar cualquier cliente, convertir cualquier moneda. Mejor escribir esa fórmula una sola vez, dejando como opcionales los datos que casi siempre son los mismos (como una tasa de IVA estándar), que copiarla y pegarla cada vez.

````python
def calcular_total_con_iva(precio_unitario, unidades, iva=0.19):
    """Calcula el total de una venta incluyendo el IVA."""
    subtotal = precio_unitario * unidades
    total = subtotal * (1 + iva)
    return total

# Reutilizamos la misma función para varias ventas distintas
venta_1 = calcular_total_con_iva(8500, 37)
venta_2 = calcular_total_con_iva(120000, 2)
venta_3 = calcular_total_con_iva(15000, 10, iva=0.05)  # un producto con IVA reducido

print("Venta 1 con IVA:", venta_1)
print("Venta 2 con IVA:", venta_2)
print("Venta 3 con IVA (5%):", venta_3)
````

Fíjate en la definición de `calcular_total_con_iva`: tiene tres parámetros, pero no los tres funcionan igual.

`precio_unitario` y `unidades` son parámetros obligatorios: no tienen un valor por defecto, así que hay que dárselos siempre que se llama la función. Si falta alguno, Python responde con un `TypeError` indicando qué argumento hizo falta.

`iva`, en cambio, es un parámetro opcional, porque en su definición ya trae un valor por defecto (`iva=0.19`). Si no lo mencionas al llamar la función, Python asume ese 19% automáticamente; si sí lo mencionas, tu valor reemplaza al de por defecto. Esto es justo lo que pasó en `venta_3`, donde se indicó `iva=0.05` para un producto con una tarifa distinta.

````python
# Si no le pasamos iva, se usa el valor por defecto (0.19)
print(calcular_total_con_iva(10000, 5))

# Si sí lo indicamos, ese valor reemplaza al de por defecto
print(calcular_total_con_iva(10000, 5, iva=0.05))

# También se pueden nombrar los parámetros obligatorios, en cualquier orden
print(calcular_total_con_iva(unidades=5, precio_unitario=10000))
````

## 9. Condicionales (`if` / `elif` / `else`)

Los condicionales permiten que el programa tome decisiones: ejecutar un bloque de código u otro según si una condición es verdadera o falsa.

```python
if condicion:
    # se ejecuta si condicion es True
elif otra_condicion:
    # se ejecuta si la primera fue False y esta es True
else:
    # se ejecuta si ninguna condición anterior fue True
```

Cualquier regla de negocio del tipo "si... entonces..." se traduce directo a un condicional: aprobar o no un crédito, clasificar un cliente, aplicar un descuento, encender una alerta de inventario bajo. También sirve para clasificar un país como en recesión si el crecimiento del PIB es negativo, decidir si un empleado recibe una alerta de desempeño, o clasificar una venta como alta, media o baja según su monto.

````python
def clasificar_venta(monto):
    if monto >= 1000000:
        return "alta"
    elif monto >= 300000:
        return "media"
    else:
        return "baja"

print("Venta de", 1500000, "se clasifica como:", clasificar_venta(1500000))
print("Venta de", 500000, "se clasifica como:", clasificar_venta(500000))
print("Venta de", 90000, "se clasifica como:", clasificar_venta(90000))
````

## 10. Ciclos (`for` y `while`)

Los ciclos permiten repetir una acción muchas veces sin escribir el mismo código una y otra vez. Hay dos formas principales.

`for elemento in iterable:` recorre, uno por uno, cada elemento de un **iterable**, que es el nombre general que se le da en Python a cualquier colección que se pueda recorrer de a un elemento a la vez. Las listas, las tuplas y los textos son ejemplos de iterables (y además se pueden indexar, como vimos en las secciones anteriores), pero no son los únicos: el diccionario que vimos hace un momento también es iterable con `for`, aunque no se indexe con números como una lista. A continuación se ve el mismo patrón de `for` aplicado, en celdas separadas y solo como ejemplo, a una lista, a una tupla y a un texto.

`while condicion:` repite un bloque de código mientras la condición siga siendo verdadera. No necesita de antemano una secuencia para recorrer, así que es la opción natural cuando no se sabe cuántas veces hay que repetir algo, sino que eso depende de que se cumpla o no una condición.

````python
# for sobre una lista
ventas_semana = [820, 950, 700, 1100]

for venta in ventas_semana:
    print("Venta registrada:", venta)
````

````python
# for sobre una tupla: se recorre exactamente igual que la lista
trimestres = ("Q1", "Q2", "Q3", "Q4")

for trimestre in trimestres:
    print("Generando reporte del", trimestre)
````

````python
# for sobre un texto: cada vuelta entrega un caracter
codigo = "AB12"

for caracter in codigo:
    print(caracter)
````

Fíjate que el `for` funcionó exactamente igual en los tres casos: lo único que cambió fue el iterable que se puso después de `in` (una lista, una tupla, un texto). Esa es la idea central: `for` no es un comando "para listas", es un comando "para iterables", y una lista es apenas uno de varios tipos de datos que cumplen esa condición.

Cuando no queremos recorrer una colección que ya existe, sino simplemente repetir algo un número determinado de veces o generar una secuencia de números, usamos `range()`. `range(n)` genera los números desde 0 hasta n-1. `range(inicio, fin)` genera desde `inicio` hasta `fin - 1`. Y `range(inicio, fin, paso)` permite avanzar de a `paso` en `paso`, en vez de uno en uno.

````python
# range(n): de 0 a n-1
for i in range(5):
    print("Iteración número", i)

# range(inicio, fin): útil para simular, por ejemplo, los 12 meses del año
for mes in range(1, 13):
    print("Mes", mes)

# range(inicio, fin, paso): generar los números de factura de 2 en 2
for numero_factura in range(1001, 1011, 2):
    print("Factura número", numero_factura)
````

## 11. Lectura básica de errores

Equivocarse escribiendo código es normal, hasta el desarrollador más experimentado lo hace todos los días. Lo importante es aprender a leer el mensaje de error (traceback) en vez de asustarse. Ya vimos dos ejemplos reales, un `TypeError` al intentar modificar una tupla y otro al llamarle una función sin uno de sus argumentos obligatorios. Estos son otros de los más comunes:

| Error | Cuándo aparece | Ejemplo |
|---|---|---|
| `SyntaxError` | La sintaxis está mal escrita (falta `:`, paréntesis sin cerrar, etc.) | `if x > 5` (sin `:`) |
| `NameError` | Usas una variable que no existe o está mal escrita | `print(precio_unitrio)` |
| `TypeError` | Mezclas tipos de datos incompatibles, o llamas mal una función | `"5" + 5` |
| `IndexError` | Pides una posición que no existe en una secuencia | `lista[100]` en una lista de 3 elementos |
| `KeyError` | Buscas una clave que no existe en un diccionario | `precios["Impresora"]` si esa clave no está |
| `ZeroDivisionError` | Divides entre cero | `10 / 0` |

Python casi siempre indica en qué línea ocurrió el problema y qué tipo de error es, en la última línea del mensaje. Conviene leerlo de abajo hacia arriba: la última línea dice qué pasó, las líneas anteriores dicen dónde. Saber leer errores rápido ahorra horas de trabajo: en vez de esperar a que alguien más "te arregle el código", puedes identificar si el problema es un nombre mal escrito, un tipo de dato incorrecto o una posición fuera de rango.

````python
# Ejecuta esta celda para ver un TypeError adicional y practicar su lectura
precio_texto = "8500"
unidades = 10

# Esto falla: no se puede sumar directamente un texto (str) con un número (int)
total = precio_texto + unidades
print(total)
````

Fíjate que la última línea del error dice algo como `TypeError: can only concatenate str (not "int") to str`. Eso dice exactamente el problema: Python no sabe sumar un `str` con un `int` directamente. La solución es convertir el texto a número con `int()` o `float()` antes de operar.

````python
# Versión corregida: convertimos el texto a número antes de operar
precio_texto = "8500"
unidades = 10

total = int(precio_texto) * unidades   # convertimos "8500" a 8500 con int()
print("Total corregido:", total)
````

Vamos a combinar todo lo que vimos hoy con un caso simple de negocio: una pequeña empresa registró sus ventas de la semana como una lista de diccionarios, donde cada diccionario representa una venta con sus propios campos: producto, monto y vendedor (por ejemplo, `{"producto": "Cuaderno", "monto": 85000, "vendedor": "Ana"}`). Para armar el total por vendedor vamos a apoyarnos en lo que ya vimos sobre diccionarios: partir de uno vacío, usar `in` / `not in` para saber si una llave ya existe, y recorrerlo al final con `.items()`.

````python
# 1) "Cargar" los datos (en la vida real vendrían de un archivo; hoy los escribimos a mano)
ventas = [
    {"producto": "Cuaderno", "monto": 85000, "vendedor": "Ana"},
    {"producto": "Mochila", "monto": 320000, "vendedor": "Luis"},
    {"producto": "Lapicero", "monto": 15000, "vendedor": "Ana"},
    {"producto": "Calculadora", "monto": 210000, "vendedor": "Marta"},
    {"producto": "Cuaderno", "monto": 85000, "vendedor": "Luis"},
    {"producto": "Mochila", "monto": 320000, "vendedor": "Marta"},
]

# 2) Inspeccionar: ¿cuántos registros tenemos? ¿cómo se ve el primero?
print("Número de ventas registradas:", len(ventas))
print("Primera venta:", ventas[0])
print("Monto de la primera venta:", ventas[0]["monto"])
````

````python
# 3) Transformar: calcular el total vendido y el total por vendedor
total_general = 0
total_por_vendedor = {}   # dict vacío: iremos acumulando aquí

for venta in ventas:
    total_general += venta["monto"]

    vendedor = venta["vendedor"]
    if vendedor not in total_por_vendedor:
        total_por_vendedor[vendedor] = 0
    total_por_vendedor[vendedor] += venta["monto"]

print("Total general de la semana:", total_general)
print("Total por vendedor:", total_por_vendedor)
````
