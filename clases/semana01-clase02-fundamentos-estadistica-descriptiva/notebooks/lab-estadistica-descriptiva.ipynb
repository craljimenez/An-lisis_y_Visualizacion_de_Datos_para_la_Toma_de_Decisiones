## 3. Medidas de tendencia central

Vamos a calcular media, mediana y moda de `total_bill` (la cuenta total).

```python
media = tips["total_bill"].mean()
mediana = tips["total_bill"].median()
moda_size = tips["size"].mode()[0]  # la moda tiene más sentido en 'size' (número de comensales)

print(f"Media de la cuenta total:   {media:.2f} USD")
print(f"Mediana de la cuenta total: {mediana:.2f} USD")
print(f"Moda del tamaño de grupo:   {moda_size} personas")
```

## 4. Medidas de dispersión

Ahora rango, varianza y desviación estándar.

```python
rango = tips["total_bill"].max() - tips["total_bill"].min()
varianza = tips["total_bill"].var()
desviacion = tips["total_bill"].std()

print(f"Rango:               {rango:.2f} USD")
print(f"Varianza:            {varianza:.2f} USD^2")
print(f"Desviación estándar: {desviacion:.2f} USD")
```

## 5. Percentiles y detección inicial de valores atípicos

Usamos el criterio del **rango intercuartílico (IQR)**: un valor es
atípico si está muy por debajo de P25 o muy por encima de P75 (más de
1.5 veces el IQR).

```python
p25 = tips["total_bill"].quantile(0.25)
p50 = tips["total_bill"].quantile(0.50)
p75 = tips["total_bill"].quantile(0.75)
iqr = p75 - p25

limite_superior = p75 + 1.5 * iqr
limite_inferior = p25 - 1.5 * iqr

print(f"P25={p25:.2f}  P50={p50:.2f}  P75={p75:.2f}  IQR={iqr:.2f}")
print(f"Límite superior para atípicos: {limite_superior:.2f} USD")
print(f"Límite inferior para atípicos: {limite_inferior:.2f} USD")
```

```python
atipicos = tips[(tips["total_bill"] > limite_superior) | (tips["total_bill"] < limite_inferior)]
print(f"Número de cuentas atípicas: {len(atipicos)}")
atipicos[["total_bill", "tip", "size", "day"]]
```
