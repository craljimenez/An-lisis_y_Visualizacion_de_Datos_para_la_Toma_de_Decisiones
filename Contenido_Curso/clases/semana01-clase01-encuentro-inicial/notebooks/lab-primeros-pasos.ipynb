## 2. Nuestra primera tabla de datos

Esta es una mini "encuesta de bienvenida" con datos **ficticios** de un grupo
parecido al nuestro. La construimos directamente en el notebook (no
necesitamos subir ningún archivo).

```python
datos = {
    "estudiante": [f"E{i:02d}" for i in range(1, 21)],
    "horas_sueno": [6, 7, 5, 8, 6, 7, 4, 6, 7, 9, 5, 6, 7, 8, 6, 5, 7, 6, 8, 7],
    "tazas_cafe": [2, 1, 3, 0, 2, 1, 4, 2, 1, 0, 3, 2, 1, 0, 2, 3, 1, 2, 0, 1],
    "herramienta_favorita": [
        "Excel", "Python", "Excel", "Power BI", "Python", "Excel", "SQL",
        "Python", "Power BI", "Excel", "SQL", "Python", "Excel", "Power BI",
        "Python", "SQL", "Excel", "Python", "Power BI", "Excel",
    ],
    "ciudad": [
        "Medell\u00edn", "Bogot\u00e1", "Medell\u00edn", "Cali", "Medell\u00edn",
        "Bogot\u00e1", "Medell\u00edn", "Cali", "Medell\u00edn", "Bogot\u00e1",
        "Medell\u00edn", "Cali", "Bogot\u00e1", "Medell\u00edn", "Medell\u00edn",
        "Cali", "Bogot\u00e1", "Medell\u00edn", "Medell\u00edn", "Cali",
    ],
}

df = pd.DataFrame(datos)
df.head()
```

## 4. Primeros estadísticos descriptivos

Esto es solo un adelanto — lo veremos a fondo en la **Clase 3**. Por ahora,
quedémonos con la idea general: la **media** y la **mediana** resumen en un
solo número el "centro" de una variable.

```python
media_sueno = df["horas_sueno"].mean()
mediana_sueno = df["horas_sueno"].median()

print(f"Media de horas de sue\u00f1o:   {media_sueno:.2f}")
print(f"Mediana de horas de sue\u00f1o: {mediana_sueno:.2f}")
```

## 6. Nuestro primer gráfico

Un gráfico de barras con la cantidad de estudiantes por herramienta favorita.

```python
conteo = df["herramienta_favorita"].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(conteo.index, conteo.values, color="#0CB3B6")
ax.set_title("Herramienta favorita del grupo")
ax.set_ylabel("N\u00famero de estudiantes")
plt.show()
```
